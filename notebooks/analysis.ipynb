# Paper-run analysis

These cells read JSONL the simulator already wrote. They do not place orders.

The committed charts under `reports/sample/` come from a **40-minute synthetic replay**, not from a two-week paper run. Record real Binance data with `make record`, then replay those gzip files. `vol_vs_24h` stays empty until the tape covers a full day. Binance spot has no gas; an earlier sample charged fake gas and is obsolete.


```python
from pathlib import Path
from jev_paper.metrics import STRATEGIES, calibration, latency_histogram, load_logs, regime_breakdown, stale_rate, strategy_summary

LOG_DIR = Path('logs/sample')
logs = load_logs(LOG_DIR)
for name in STRATEGIES:
    print(name, strategy_summary(logs[name]))
print('stale', stale_rate(logs['decisions']))
print('calibration', calibration(logs['decisions'], logs['snapshots']))
print('regimes', regime_breakdown(logs['decisions'], logs['outcomes']))
print('latency', latency_histogram(logs['decisions']))
```
